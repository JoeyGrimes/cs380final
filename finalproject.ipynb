### 1. Data Preprocessing

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
#load the data first
import re



def clean_for_tfidf(line):
    line = line.lower()
    line = re.sub(r'\[url\]', '', line)
    line = re.sub(r'\[user\]', '', line)
    line = re.sub(r'\s+', ' ', line).strip()
    return line

def clean_for_roberta(line):
    line = re.sub(r'\[url\]', '', line)
    line = re.sub(r'\[user\]', '', line)
    line = re.sub(r'\s+', ' ', line).strip()
    return line

df = pd.read_csv('edos_labelled_data.csv')

df['idf_cleaned'] = df['text'].apply(clean_for_tfidf)
df['clean_text'] = df['text'].apply(clean_for_roberta)

#clean the test
train_df = df[df['split'] == 'train']
test_df = df[df['split'] == 'test']

X_train_idf = train_df['idf_cleaned']
Y_train_label = train_df['label']
X_test_idf = test_df['idf_cleaned']
Y_test_label = test_df['label']
    
```

### 2. Feature Engineering

```python
# TF-IDF Setup
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(ngram_range=(1,2), sublinear_tf=True)
X_train_tfidf_fitted = tfidf.fit_transform(X_train_idf)
X_test_tfidf_fitted = tfidf.transform(X_test_idf)
```

```python
# roBERTA setup
import torch
from transformers import AutoTokenizer, AutoModel
import numpy as np

# This is loading th roberta base
MODEL_NAME="cardiffnlp/twitter-roberta-base-hate"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModel.from_pretrained(MODEL_NAME)
model.eval()

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

def getEmbeddings(texts, batchSize=32):
    # Mean-pool last hidden state to get the sentence embeddings
    allEmbeddings = []

    for i in range(0, len(texts), batchSize):
        batch = texts[i:i+batchSize]
        encoded = tokenizer(
            list(batch),
            padding=True,
            truncation=True,
            max_length=128,
            return_tensors = "pt"
        ).to(device)

        with torch.no_grad():
            output = model(**encoded)

        # Mean pool over token dimension

        attention_mask = encoded["attention_mask"]
        token_embeddings = output.last_hidden_state # (batch, sequence_length, 768) i dont know what this is for yet
        mask_expanded = attention_mask.unsqueeze(-1).float()
        embeddings = (token_embeddings * mask_expanded).sum(1) / mask_expanded.sum(1)

        allEmbeddings.append(embeddings.cpu().numpy())

        if i % 200 == 0:
            print(f"   Processed {i}/{len(texts)}")

    return np.vstack(allEmbeddings)

X_train_roberta = getEmbeddings(train_df["clean_text"].values)
X_test_roberta = getEmbeddings(test_df["clean_text"].values)
print(f"RoBERTa embedding shape: {X_train_roberta.shape}")
```

### 3. Model Selection and Architecture


```python
# TF-IDF Logistic Regression CV 0.77
from sklearn.linear_model import LogisticRegressionCV
from sklearn.metrics import classification_report

lr = LogisticRegressionCV(cv=5, max_iter=1000, class_weight='balanced', random_state=42)
lr.fit(X_train_tfidf_fitted, Y_train_label)
y_pred_lr = lr.predict(X_test_tfidf_fitted)

print(classification_report(Y_test_label, y_pred_lr))
```

```python
# TF-IDF SVM 0.79
from sklearn.svm import LinearSVC
from sklearn.metrics import classification_report

svm = LinearSVC(C=10, max_iter=100000, class_weight='balanced', random_state=42)
svm.fit(X_train_tfidf_fitted, Y_train_label)
y_pred_svm = svm.predict(X_test_tfidf_fitted)

print(classification_report(Y_test_label, y_pred_svm))
```

```python
# TF-IDF XGBoost 0.78
import xgboost as xgb
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
Y_train_encoded = le.fit_transform(Y_train_label)
Y_test_encoded = le.transform(Y_test_label)

xgb_model = xgb.XGBClassifier(
    n_estimators=400, 
    max_depth=6, 
    learning_rate=0.1, 
    random_state=42, 
    eval_metric='logloss'
)

xgb_model.fit(X_train_tfidf_fitted, Y_train_encoded)
y_pred_enc = xgb_model.predict(X_test_tfidf_fitted)
y_pred_xgb = le.inverse_transform(y_pred_enc)

print(classification_report(Y_test_label, y_pred_xgb))
```

```python
# Roberta - SVM 0.76
from sklearn.svm import LinearSVC

svm_roberta = LinearSVC(C=10, max_iter=5000, class_weight='balanced', random_state=42)
svm_roberta.fit(X_train_roberta, Y_train_label)
y_pred_svm_roberta = svm_roberta.predict(X_test_roberta)

print(classification_report(Y_test_label, y_pred_svm_roberta))
```

```python
# Roberta XGBoost - a bit better f1 weighted at .81
import xgboost as xgb
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report

le = LabelEncoder()
Y_train_encoded = le.fit_transform(Y_train_label)
Y_test_encoded = le.transform(Y_test_label)

# if I adjust any of these params it loses accuracy, best still at .81
xgb_roberta = xgb.XGBClassifier(
    n_estimators=500,
    max_depth=6,
    learning_rate=0.05,
    random_state=42,
    eval_metric='logloss'
)

xgb_roberta.fit(X_train_roberta, Y_train_encoded)
y_pred_enc = xgb_roberta.predict(X_test_roberta)
y_pred_roberta_xgb = le.inverse_transform(y_pred_enc)

print(classification_report(Y_test_label, y_pred_roberta_xgb))
```

```python
# Need the .backward() so the weights can actually self adjust from the GD. This takes roughly 50 minutes for 3 epochs but yielded weighted f1 avg ~ 0.85
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
import torch

MODEL_NAME = "cardiffnlp/twitter-roberta-base-hate"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# Dataset class
class SexismDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(list(texts), truncation=True, padding=True, max_length=128)
        self.labels = []
        for l in labels:
            if l == 'sexist':
                self.labels.append(1)
            else:
                self.labels.append(0)

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        item = {}
        for k, v in self.encodings.items():
            item[k] = torch.tensor(v[idx])
        item['labels'] = torch.tensor(self.labels[idx])
        return item

# Create datasets
train_dataset = SexismDataset(train_df['clean_text'].values, Y_train_label.values)
test_dataset  = SexismDataset(test_df['clean_text'].values, Y_test_label.values)

#split data into chunks to feed to GPU
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader  = DataLoader(test_dataset, batch_size=16)

# Loads pretrained roberta weight from twitter 
model_ft = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=2)
model_ft = model_ft.to(device)

optimizer = AdamW(model_ft.parameters(), lr=2e-5)

# Training loop
for epoch in range(3):
    model_ft.train()
    total_loss = 0
    for batch in train_loader:
        optimizer.zero_grad()
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels = batch['labels'].to(device)

        outputs = model_ft(input_ids=input_ids, attention_mask=attention_mask, labels=labels)
        loss = outputs.loss
        loss.backward()
        # update weights with optimizer after calculating the loss
        optimizer.step()
        total_loss += loss.item()

    print(f"Epoch {epoch+1} loss: {total_loss/len(train_loader):.4f}")

# Evaluation
model_ft.eval()
all_preds = []
with torch.no_grad():
    for batch in test_loader:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        outputs = model_ft(input_ids=input_ids, attention_mask=attention_mask)
        preds = torch.argmax(outputs.logits, dim=1)
        all_preds.extend(preds.cpu().numpy())

y_pred_finetuned = []
for p in all_preds:
    if p == 1:
        y_pred_finetuned.append('sexist')
    else:
        y_pred_finetuned.append('not sexist')
print(classification_report(Y_test_label, y_pred_finetuned))
```

 Evaluation and Results

```python
import pandas as pd
from sklearn.metrics import classification_report

# Format: (Model Name, True Labels, Predicted Labels)
experiments = [
    ("TF-IDF + Logistic Regression", Y_test_label, y_pred_lr),
    ("TF-IDF + Linear SVM", Y_test_label, y_pred_svm),
    ("TF-IDF + XGBoost", Y_test_label, y_pred_xgb),
    ("RoBERTa + Linear SVM", Y_test_label, y_pred_svm_roberta),
    ("RoBERTa + XGBoost", Y_test_label, y_pred_roberta_xgb),
    ("RoBERTa Fine-tuned (Best)", Y_test_label, y_pred_finetuned)
]

table_rows = []

for name, y_true, y_pred in experiments:
  
    report = classification_report(y_true, y_pred, output_dict=True)
    
    row = {
        'Feature + Model': name,
        'Sexist (P)': report['sexist']['precision'],
        'Sexist (R)': report['sexist']['recall'],
        'Sexist (F1)': report['sexist']['f1-score'],
        'Non-Sexist (P)': report['not sexist']['precision'],
        'Non-Sexist (R)': report['not sexist']['recall'],
        'Non-Sexist (F1)': report['not sexist']['f1-score'],
        'Weighted (P)': report['weighted avg']['precision'],
        'Weighted (R)': report['weighted avg']['recall'],
        'Weighted (F1)': report['weighted avg']['f1-score']
    }
    table_rows.append(row)

results_df = pd.DataFrame(table_rows)

# Rounding here so it's easier to read
results_df = results_df.round(3)

display(results_df)

best_model_row = results_df.loc[results_df['Weighted (F1)'].idxmax()]
print(f"\nBest performer: {best_model_row['Feature + Model']} with a Weighted F1 of {best_model_row['Weighted (F1)']}")
```
